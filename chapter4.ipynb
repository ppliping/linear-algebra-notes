# 第四章 —— 向量空间

---
# 第一部分　知识点总结

## 4.1　向量空间与子空间

- **向量空间（vector space）**：一个带有加法与标量乘法两种运算的集合 $V$，元素统称**向量（vectors）**，运算满足以下 **10 条公理**（对 $V$ 中所有 $\mathbf u,\mathbf v,\mathbf w$ 与所有标量 $c,d$）：
  1. $\mathbf u+\mathbf v\in V$（加法封闭）；
  2. $\mathbf u+\mathbf v=\mathbf v+\mathbf u$；
  3. $(\mathbf u+\mathbf v)+\mathbf w=\mathbf u+(\mathbf v+\mathbf w)$；
  4. $V$ 中存在**零向量** $\mathbf 0$：$\mathbf u+\mathbf 0=\mathbf u$；
  5. 对每个 $\mathbf u$ 存在**负向量** $-\mathbf u$：$\mathbf u+(-\mathbf u)=\mathbf 0$；
  6. $c\mathbf u\in V$（数乘封闭）；
  7. $c(\mathbf u+\mathbf v)=c\mathbf u+c\mathbf v$；
  8. $(c+d)\mathbf u=c\mathbf u+d\mathbf u$；
  9. $c(d\mathbf u)=(cd)\mathbf u$；
  10. $1\mathbf u=\mathbf u$。
- **例子**：
  - $\mathbb R^n$（$n\ge1$）；三维空间里的箭头
  - **（离散时间）信号空间** $S$：全体（双向无穷）数列 $\{y_k\}$（$k$ 取遍全体整数），加法与数乘按**逐项**进行
  - **多项式空间** $\mathbb P_n$：次数 $\le n$ 的全体多项式
    $p(t)=a_0+a_1t+\cdots+a_nt^n$
    （零多项式也属于 $\mathbb P_n$，其次数按约定不作定义）
  - 定义在集合 $D$ 上的全体实值函数（按点态加法、数乘）。把每个函数看成“一个向量”是本章的基本观点
  - 全体 $m\times n$ 矩阵 $M_{m\times n}$（按元素加、数乘）
- **子空间（subspace）**：$V$ 的子集 $H$ 若满足三条性质就是 $V$ 的子空间：
  1. $\mathbf 0\in H$；
  2. 对加法封闭；
  3. 对数乘封闭。
  
  此时 $H$ 自动满足全部 10 条公理（其余公理对 $V$ 成立，自然对 $H$ 成立），故 $H$ 本身是向量空间。**检验顺序**：先找零向量，$\mathbf 0\notin H$ 则立即排除。
- **例子**：
  1. 只含 $\mathbf 0$ 的**零子空间** $\{\mathbf 0\}$；
  2. $\mathbb P_n$ 是全体多项式空间 $\mathbb P$ 的子空间；
  3. **有限支撑信号** $S_f$（只有有限个非零项的信号）是 $S$ 的子空间；
  4. $\mathbb R^2$ **不是** $\mathbb R^3$ 的子空间（甚至不是子集）；$\mathbb R^3$ 的非零真子空间只有**过原点的平面**（$\operatorname{Span}\{\mathbf v_1,\mathbf v_2\}$，$\mathbf v_1,\mathbf v_2$ 无关）与**过原点的直线**（$\operatorname{Span}\{\mathbf v\}$，$\mathbf v\neq\mathbf 0$）。
  5. 不过原点的平面/直线不是子空间（不含零向量）。
- **定理 1**：若 $\mathbf v_1,\dots,\mathbf v_p\in V$，则 $\operatorname{Span}\{\mathbf v_1,\dots,\mathbf v_p\}$ 是 $V$ 的子空间，称其为**由  $\{\mathbf v_1,\dots,\mathbf v_p\}$张成（生成）的子空间**；若 $H=\operatorname{Span}\{\mathbf v_1,\dots,\mathbf v_p\}$，则称  $\{\mathbf v_1,\dots,\mathbf v_p\}$为 $H$ 的一个**张成集（生成集）**。
  - 把集合 $H$ 改写成若干向量的线性组合（即 $H=\operatorname{Span}\{\cdots\}$），用定理 1 判定它是子空间。

## 4.2　零空间、列空间、行空间与线性变换

- **零空间(null space)**：$m\times n$ 矩阵 $A$ 的零空间
  $$\operatorname{Nul}A=\{\mathbf x:\mathbf x\in\mathbb R^n and \ A\mathbf x=\mathbf 0\}$$
  即齐次方程 $A\mathbf x=\mathbf 0$ 的全体解，也是被映射 $\mathbf x\mapsto A\mathbf x$ 送到零向量的全体 $\mathbf x$。
  - **定理 2**：$\operatorname{Nul}A$ 是 $\mathbb R^n$ 的子空间
  - **显式描述**：解 $A\mathbf x=\mathbf 0$（行化简 + 参数向量形式）就得到 $\operatorname{Nul}A$ 的显式描述。所得张成向量自动线性无关（自由变量作权重），且**张成向量的个数 = 自由变量个数**。
- **列空间(column space)**：$m\times n$ 矩阵 $A = [a_1 \dots a_n]$ 的列空间
  $$\operatorname{Col}A=\operatorname{Span}\{\mathbf a_1,\dots,\mathbf a_n\}=\{\mathbf b:\ \mathbf b=A\mathbf x\ \text{对某个}\ \mathbf x\in\mathbb R^n\}$$
  - **定理 3**：$\operatorname{Col}A$ 是 $\mathbb R^m$ 的子空间。
  - 判定：$\mathbf b\in\operatorname{Col}A\iff$ $A\mathbf x=\mathbf b$ 相容。
- **行空间(row space)**：$A$ 的各行（视为 $\mathbb R^n$ 中向量）张成的子空间
   $$\operatorname{Row}A = \operatorname Span\{r_1 \dots r_m \}$$
  - 是 $\mathbb R^n$ 的子空间；$\operatorname{Row}A=\operatorname{Col}A^T$。
- **$\operatorname{Nul}A$ 与 $\operatorname{Col}A$ 的对照**（$A$ 为 $m\times n$）：
  | | $\operatorname{Nul}A$ | $\operatorname{Col}A$ |
  |---|---|---|
  | 所属空间 | $\mathbb R^n$ | $\mathbb R^m$ |
  | 定义方式 | 隐式（条件 $A\mathbf x=\mathbf 0$） | 显式（列的线性组合） |
  | 找向量 | 需行化简 $[\,A\ \mathbf 0\,]$ | 直接用 $A$ 的列 |
  | 典型判据 | $A\mathbf v=\mathbf 0$ | $ A \mathbf x = \mathbf v$ 相容 |
  | 映射 | $\operatorname{Nul}A=\{\mathbf 0\}\iff A\mathbf x=\mathbf 0$ 只有零解 $\iff$ $\mathbf x\mapsto A\mathbf x$ 单射 | $\operatorname{Col}A=\mathbb R^m\iff A\mathbf x=\mathbf b$ 对每个 $\mathbf b$ 有解 $\iff$ 满射 |
- **线性变换$T:V\to W$**
  - 从向量空间 $V$ 到向量空间 $W$ 的线性变换 $T$ 是一条规则，它将 $V$ 中的每个向量 $\mathbf x$ 映射到 $W$ 中一个唯一的向量 $T(\mathbf x)$，使得$$T(\mathbf u+\mathbf v) = T(\mathbf u)+T(\mathbf v) \quad T(c \mathbf u) = cT(\mathbf u)$$
  - **核（kernel，零空间）** $\ker T=\{\mathbf u\in V:\ T(\mathbf u)=\mathbf 0\}$，是 $V$ 的子空间；
  - **值域（range）** $\operatorname{range}(T)=\{T(\mathbf x):\ \mathbf x\in V\}$（也称像集），是 $W$ 的子空间。
  - 对 $T(\mathbf x)=A\mathbf x$：$\ker T=\operatorname{Nul}A$，$\operatorname{range}(T)=\operatorname{Col}A$。
  - 微积分例子：
    - $D:C^1[a,b] \rightarrow C[a,b],\ \ Df = f'$ 是线性变换，$\ker D $= {常数函数}，$\operatorname{range} D =C[a,b]$
    - $L : C^2[a, b] \to C[a, b],\ \ L(y) = y'' + \omega^2 y$，是线性变换，$\ker L = \{y \in C^2[a, b] : y'' + \omega^2 y = 0\},$ 即微分方程 $y'' + \omega^2 y = 0$ 的解集就是线性映射 $L$ 的核
  - **滑动平均**（信号平滑）：$M_2\{p_k\}=\{z_k\}$，$z_k=\frac12(p_k+p_{k-1})$ 是 $S\to S$ 的线性变换，其核由交替信号 $\alpha=\{\ldots,1,-1,1,-1,\ldots\}$ 张成（$M_2$ 把“隔日震荡”抹掉）。

## 4.3　线性无关集；基

- **线性无关**：$V$ 中向量组 $\{\mathbf v_1,\dots,\mathbf v_p\}$，若向量方程
  $$c_1\mathbf v_1+\cdots+c_p\mathbf v_p=\mathbf 0$$
  只有零解 $c_1=\cdots=c_p=0$，则称线性无关；否则线性相关。
- **定理 4（线性相关判据）**：指标集 $\{\mathbf v_1,\dots,\mathbf v_p\}$（$p\ge2,\mathbf v_1 \neq 0$）线性相关 $\iff$ 存在某个 $\mathbf v_j$（$j\ge2$）是它前面向量 $\mathbf v_1,\dots,\mathbf v_{j-1}$ 的线性组合。特例：含零向量必相关；两个向量相关 $\iff$ 一个是另一个的倍数。
- **定理 5（张成集定理）**：设 $S=\{\mathbf v_1,\dots,\mathbf v_p\}$ 张成 $H$。
  - a. 若某个 $\mathbf v_k$ 是其余向量的线性组合，则去掉 $\mathbf v_k$ 后剩下的仍张成 $H$；
  - b. 若 $H\neq\{\mathbf 0\}$，则$S$ 的某个子集是 $H$的一个基。
- **基（basis）**：$H$（$V$ 的子空间）的基 $B=\{\mathbf b_1,\dots,\mathbf b_p\}$ 是 $H$ 中满足以下两条的向量组：
  1. $B$ 线性无关；
  2. $H=\operatorname{Span}B$。（$\mathbb R^n$ 的标准基(standard basis) $\{\mathbf e_1,\dots,\mathbf e_n\}$。）
- **求基的三个标准方法**：
  - $\operatorname{Nul}A$ 的基：解 $A\mathbf x=\mathbf 0$ 的参数向量形式中每个自由变量对应的向量；
  - **定理 6**：$\operatorname{Col}A$ 的基 = $A$ 的**主元列**（注意取 $A$ 本身的列，不是行化简后矩阵的列）；
  - **定理 7**：行等价的矩阵有相同的行空间；$A$ 行化简到阶梯形 $B$ 后，$B$ 的**非零行**构成 $\operatorname{Row}A$ 的基，也构成 $\operatorname{Row}B$ 的基。
- **基的两种看法**：基是 $H$ 中**尽可能小的张成集**（再删就张不成了），也是 $H$ 中**尽可能大的无关集**（再加一个 $H$ 中的向量就相关了）。

## 4.4　坐标系统
- $\mathcal B$-坐标
  - **定理 8（唯一表示定理）**：若 $\mathcal B=\{\mathbf b_1,\dots,\mathbf b_n\}$ 是 $V$ 的基，则每个 $\mathbf x\in V$ 都能**唯一**地写成基向量的线性组合
  $$\mathbf x=c_1\mathbf b_1+\cdots+c_n\mathbf b_n.$$
  - **坐标**：上式中的权重称为 $\mathbf x$ 的 **$\mathcal B$-坐标**，坐标向量
  $$[\mathbf x]_\mathcal B=\begin{bmatrix}c_1\\\vdots\\c_n\end{bmatrix}.$$
  取坐标的过程 $\mathbf x\mapsto[\mathbf x]_{\mathcal B}$ 是 $V\to\mathbb R^n$ 的**坐标映射**。

- **$\mathbb R^n$ 的情形**：若 $\mathcal B=\{\mathbf b_1,\dots,\mathbf b_n\}$ 是 $\mathbb R^n$ 的基，令
  $$P_\mathcal B=\begin{bmatrix}\mathbf b_1&\cdots&\mathbf b_n\end{bmatrix},\qquad \mathbf x=P_\mathcal B[\mathbf x]_\mathcal B,$$
  $P_\mathcal B$ 称为（由 $\mathcal B$ 确定的）**坐标变换矩阵(change-of-coordinates matrix)**，可逆，$[\mathbf x]_\mathcal B=P_\mathcal B^{-1}\mathbf x$。

- **定理 9**：设 $\mathcal B=\{\mathbf b_1,\dots,\mathbf b_n\}$ 是 $V$ 的基，则坐标映射
  1. 把 $V$ 一一映射到 $\mathbb R^n$（既单射又满射）；
  2. 是**线性变换**（保加法、保数乘）：
     $[\mathbf u+\mathbf w]_\mathcal B=[\mathbf u]_\mathcal B+[\mathbf w]_\mathcal B$，$[c\mathbf u]_B=c[\mathbf u]_\mathcal B$。
  这样的映射称为 $V$ 与 $\mathbb R^n$ 之间的**同构（isomorphism）**；$V$ 与 $\mathbb R^n$ 同构，记 $V\cong\mathbb R^n$。线性无关、线性组合、张成等性质在坐标映射下保持。

- **$\mathbb P_n$ 的情形**：取标准基 $\{1,t,\dots,t^n\}$，多项式 $p(t)=a_0+a_1t+\cdots+a_nt^n$ 的坐标向量就是系数向量 $(a_0,\dots,a_n)$。多项式的线性无关/张成问题由此化为 $\mathbb R^{n+1}$ 中的对应问题。$\rightarrow$ 坐标映射把抽象向量空间中的计算**搬到 $\mathbb R^n$ 中**用矩阵完成

## 4.5　向量空间的维数

- **维数（dimension）**：
  - 若 $V$ 由有限个向量张成，则称 $V$ **有限维(finite-dimensional)**；$V$ 的基所含向量个数称为 $V$ 的**维数** $\dim V$
    $$\dim\{\mathbf 0\}=0,\ \\dim\mathbb R^n=n, \ \\dim\mathbb P_n=n+1$$
  - 不是有限维的空间称**无限维(infinite-dimensional)**
    $$全体多项式空间 \mathbb P、信号空间 S$$
- $\mathbb R^3$ 的子空间：$\{\mathbf 0\}$（0 维）、过原点直线（1 维）、过原点平面（2 维）、$\mathbb R^3$（3 维）——再无其他。
- **定理 10**：若 $V$ 有含 $n$ 个向量的基，则 $V$ 中**多于 $n$ 个**向量的组必线性相关。
- **定理 11**：若 $V$ 有含 $n$ 个向量的基，则 $V$ 的**每个**基都恰含 $n$ 个向量
- **定理 12**：$H$ 是有限维空间 $V$ 的子空间，则 $H$ 中任何线性无关组都能（必要时）扩充成 $H$ 的基；且 $H$ 有限维，$\dim H\le\dim V$。
- **定理 13（基定理）**：$V$ 是 $p$ 维空间（$p\ge1$），则
  - $V$ 中**恰含 $p$ 个向量的线性无关组自动是基**；
  - **恰含 $p$ 个向量且张成 $V$ 的组自动是基**。
  - (验证基只需验证“无关 + 个数”或“张成 + 个数”之一)
- **秩与零化度**：对于 $m\times n$ 矩阵 $A$
  - $\operatorname{rank}A=\dim\operatorname{Col}A$ = 主元列数 = $\dim\operatorname{Row}A$
  - $\operatorname{nullity}A=\dim\operatorname{Nul}A$ = 自由变量数。
- **定理 14（秩定理）**：
  $$\operatorname{rank}A+\dim\operatorname{Nul}A=\text{（}A\text{ 的列数）}.$$
- **可逆矩阵定理**：$n\times n$ 矩阵 $A$ 可逆还等价于：
  - m. $A$ 的列构成 $\mathbb R^n$ 的基；
  - n. $\operatorname{Col}A=\mathbb R^n$；
  - o. $\operatorname{rank}A=n$；
  - p. $\operatorname{nullity}A=0$；
  - q. $\operatorname{Nul}A=\{\mathbf 0\}$。（由 $A$ 可逆 $\Rightarrow A^T$ 可逆，每条都可对 $A^T$ 复述。）
- **注记**：用行化简数主元求秩，在**精确算术**下没问题；但浮点运算会改变“表观秩”（例如 $x$ 未被精确存成 7 时 $x-7$ 是否为零取决于机器）。实际应用中常由**奇异值分解**（SVD，见 7.4 节）确定**有效秩**，并由它得到 $\operatorname{Col}A,\operatorname{Row}A,\operatorname{Nul}A,\operatorname{Nul}A^T$ 的可靠基。

## 4.6　基变换

- 问题：同一向量 $\mathbf x$ 在两组基 $B=\{\mathbf b_1,\dots,\mathbf b_n\}$、$C=\{\mathbf c_1,\dots,\mathbf c_n\}$ 下的坐标 $[\mathbf x]_B$ 与 $[\mathbf x]_C$ 如何互相换算？
- **定理 15**：设 $B,C$ 都是 $V$ 的基，则存在 $n\times n$ 矩阵 $P_{C\leftarrow B}$（**基变换矩阵 / 坐标变换矩阵**，change-of-coordinates matrix from $B$ to $C$），使
  $$[\mathbf x]_C=P_{C\leftarrow B}[\mathbf x]_B\qquad(\text{对一切}\ \mathbf x\in V),$$
  且 $P_{C\leftarrow B}$ 的列恰是 $B$ 中基向量的 $C$-坐标向量：
  $$P_{C\leftarrow B}=\begin{bmatrix}[\mathbf b_1]_C&\cdots&[\mathbf b_n]_C\end{bmatrix}.$$
  $P_{C\leftarrow B}$ 可逆，且 $$P_{B\leftarrow C}=(P_{C\leftarrow B})^{-1}$$
- **$\mathbb R^n$ 中**
  1. 按矩阵计算：把基向量排成矩阵
  $$P_B=\begin{bmatrix}\mathbf b_1&\cdots&\mathbf b_n\end{bmatrix},\qquad P_C=\begin{bmatrix}\mathbf c_1&\cdots&\mathbf c_n\end{bmatrix},$$
  则对每个 $\mathbf x\in\mathbb R^n$ 有 $\mathbf x=P_C[\mathbf x]_C$ 且 $\mathbf x=P_B[\mathbf x]_B$，故
  $$[\mathbf x]_C={P_C}^{-1}P_B[\mathbf x]_B,\qquad\boxed{P_{C\leftarrow B}=P_C^{-1}P_B}.$$
  2. 初等行变换
  $$ \boxed{[c_1 \dots c_n | b_1 \dots b_n] \sim [I\ \ P_{C\leftarrow B}]}$$
     
- **多组基的复合**：$[\mathbf x]_D=P_{D\leftarrow C}[\mathbf x]_C=P_{D\leftarrow C}P_{C\leftarrow B}[\mathbf x]_B$，故
  $$P_{D\leftarrow B}=P_{D\leftarrow C}\,P_{C\leftarrow B}$$

## 4.7　数字信号处理

- **离散时间信号（Discrete-Time Signals）的向量空间 $\mathbb S$**：全体双向无穷数列 $\{y_k\}$（$k\in\mathbb Z$）。
  - 常用信号：
  - **单位冲激（delta）** $\delta=\{d_k\}$（$d_0=1$，其余为 0）
  - **单位阶跃** $\upsilon$（$k\ge0$ 时 1，否则 0）
  - **交替信号** $\alpha=\{\ldots,1,-1,1,\ldots\}$（$(-1)^k$）
  - 指数信号 $\{a^k\}$
  - 幂信号 $\{k^n\}$
  - 正弦型信号 $\{\cos(fk\pi+\theta\pi)\}$
- **移位变换**：$S\{x_k\}=\{x_{k-1}\}$（右移一位）, $S^{-1}\{x_k\}=\{x_{k+1}\}$（左移）。$S$ 可逆且保持“时间平移”结构。
- **滑动平均（moving average）**：
  $$M_m\{x_k\}=\{y_k\},\qquad y_k=\frac1m\sum_{j=k-m+1}^{k}x_j,$$
  用于**平滑/滤噪**（如股价的 $m$ 日均线）；$m$ 越大平滑越强。LTI 的**核(kernel)**刻画了“被滤掉的东西”
- **线性时不变变换（Linear Time Invariant,LTI）**：$T:\mathbb S\to \mathbb S$ 若满足
  - (i) $T(\{x_k\}+\{y_k\})=T(\{x_k\})+T(\{y_k\})$；
  - (ii) $T(c\{x_k\})=cT(\{x_k\})$；
  - (iii) 若 $T\{x_k\}=\{y_k\}$,则对一切整数 $q$ 有 $T\{x_{k+q}\}=\{y_{k+q}\}$（时不变：输入整体平移，输出跟着平移）
    
  则称 $T$ 为 LTI 变换。移位 $S$、滑动平均 $M_m$ 都是 LTI。
- **定理 16**：LTI 变换是线性变换
- **有限长信号(signals of length n)空间**：$\mathbb S_n$ = 满足 $k<0$ 或 $k>n$ 时 $y_k=0$ 的信号全体。
  - **定理 17**：$S_n$ 是 $S$ 的子空间，与 $\mathbb R^{n+1}$ 同构，且
    $$B_n=\{\delta,\ S(\delta),\ S^2(\delta),\ \dots,\ S^n(\delta)\}$$
    是 $S_n$ 的基（$S^j(\delta)$ 就是第 $j$ 位为 1 的“单位脉冲”，相当于 $\mathbf e_{j+1}$）。于是 $S_n$ 中信号 $\{y_k\}\leftrightarrow(y_0,\dots,y_n)$。
- **有限支撑信号（finitely supported signals）空间** ：$\mathbb S_f=\{y_k\}$,  ${y_k}$是只有有限个非零项的信号。（并不是指信号长度有限，而是指非零的数值只有有限个。比如一个信号有 100 万个点，但只有前 3 个点不是 0，后面全是 0）
  - **定理 18**：$B_f=\{S^j(\delta):\ j\in\mathbb Z\}$ 是**无限维**空间 $\mathbb S_f$ 的基（基里面的向量有无数个，但每个有限支撑信号是**有限个** $S^j(\delta)$ 的组合，且表示唯一）。
  - 注意：$B_f$ **不能张成整个 $S$**——例如单位阶跃 $\upsilon=\sum_{j\ge0}S^j(\delta)$ 是**无穷和**，不算线性组合（线性组合只允许有限项）。

## 4.8　差分方程的应用

- **信号的线性无关（Casorati 判别法）**：对信号 $\{u_k\},\{v_k\},\{w_k\}$，若
  $$c_1u_k+c_2v_k+c_3w_k=0\quad\text{对一切 } k,$$
  则对 $k,k+1,k+2$ 也成立，得线性方程组，其系数矩阵
  $$C(k)=\begin{bmatrix}u_k&v_k&w_k\\u_{k+1}&v_{k+1}&w_{k+1}\\u_{k+2}&v_{k+2}&w_{k+2}\end{bmatrix}$$
  称为 **Casorati 矩阵**，其行列式称 **Casoratian**。**若 $C(k)$ 对至少某个 $k$ 可逆，则这些信号线性无关**（然而：Casorati 处处奇异并不能推出相关）。
- **线性差分方程（Linear Difference Equations）**：
  $$y_{k+n}+a_1y_{k+n-1}+\cdots+a_{n-1}y_{k+1}+a_ny_k=z_k\qquad(\text{对一切 } k)$$
  称为 $n$ **阶**线性差分方程（递推关系）；$z_k=0$ 时称**齐次**，否则**非齐次**。
- **找齐次方程的解：辅助方程法（auxiliary equation）**

  把 $y_k=r^k$ 代入齐次方程得
  $$r^n+a_1r^{n-1}+\cdots+a_{n-1}r+a_n=0,$$
  若辅助方程有 $n$ 个互异实根 $r_1, \dots, r_n$，则 $\{r_i^k\}$ 是实解空间的一组基。若有复根，实解需用 $e^{\alpha k} \cos(\beta k)$、$e^{\alpha k} \sin(\beta k)$ 表示；若有重根，还要加入 $k r^k, k^2 r^k, \dots$ 形式的解。
- **定理 19（初值问题唯一可解）**：若 $a_n\neq0$ 并给定 $z_k$ 与初值 $y_0,\dots,y_{n-1}$，则差分方程有**唯一**解（递推可以向前也可以向后无限进行）
- **定理 20**：（以下默认$a_n\neq 0$ 并在双向无限序列空间中讨论）$n$ 阶齐次线性差分方程的全体解 $H$ 是 $n$ 维向量空间。于是：找到 $n$ 个线性无关的解，由**基定理**它们自动是解空间$H$的基， 这组基叫做基本解集（fundamental set of solutions）,$H$ 是 $\mathbb S$ 的子空间（线性变换的核）。
- **非齐次方程的通解**：$\{z_k\}\neq 0$ 时，取一个特解 $y_k^{(p)}$，则解集 = 特解 + 齐次解空间：
  $$\{y_k\}=y_k^{(p)}+c_1\{r_1^k\}+\cdots+c_n\{r_n^k\}$$
 （解集是 $y^{(p)}+\operatorname{Span}\{\cdots\}$，几何上是平移过来的子空间。
- **把高阶齐次差分方程化为一阶方程组(First-Order Equations)**
$$\mathbf{x}_{k+1} = A\mathbf{x}_{k}$$
$$ \begin{bmatrix} 
y_{k+1} \\ 
y_{k+2} \\ 
\vdots \\ 
y_{k+n} 
\end{bmatrix}
 = \begin{bmatrix} 
0 & 1 & 0 & \cdots & 0 \\ 
0 & 0 & 1 & \cdots & 0 \\ 
\vdots & & & \ddots & \vdots \\ 
0 & 0 & 0 & \cdots & 1 \\ 
-a_n & -a_{n-1} & -a_{n-2} & \cdots & -a_1 
\end{bmatrix}
\begin{bmatrix} 
y_k \\ 
y_{k+1} \\ 
\vdots \\ 
y_{k+n-1} 
\end{bmatrix}$$

对非齐次方程，系统为$$\mathbf{x}_{k+1} = A\mathbf{x}_{k}+(0,\dots ,0,z_k)^T$$

**题目**：设

典型信号形如

 

证明 $W$ 是 $\mathbb S$ 的子空间。

**解答**：

$W$ 中信号由一个实数 $r$ 完全确定

设 $$\mathbf r,\,\mathbf s \in W$$

有
$$\mathbf r =\{x_k\}\ \ x_k=\begin{cases}0,&k\ \text{是}\ 2\ \text{的倍数},\\ r,&k\ \text{不是}\ 2\ \text{的倍数},\end{cases}\quad r\ \text{为任意实数}$$

- 零向量：$r=0$ 给出零信号 $\{0\}\in W$；
- 加法封闭：$\mathbf r+\mathbf s$ 在奇数位取 $r+s$、偶数位取 $0$，即 $\mathbf r+\mathbf s\in W$；

### 习题 4.7 · 第 30 题

**题目**：设
$$W=\Big\{\{x_k\}\ \Big|\ x_k=\begin{cases}0,&k<0,\\ r_k,&k\ge0,\end{cases}\quad\text{各 } r_k\ \text{为任意实数}\Big\},$$
典型信号形如



证明 $W$ 是 $S$ 的子空间。

**解答**：

$W$ 是“负半轴全为零”的信号全体

- 零向量：取所有 $r_k=0$ 得零信号 $\in W$；
- 加法封闭：$\{x_k\},\{y_k\}\in W$，则 $k<0$ 时 $(x_k+y_k)=(x+y)_k=0+0=0$，故 $\{x_k+y_k\}\in W$；
- 数乘封闭：$k<0$ 时 $cx_k = (cx)_k=c\cdot0=0$，故 $c\{x_k\}\in W$。

故 $W$ 是 $S$ 的子空间

### 习题 4.7 · 第 32 题

**题目**：描述第 30 题的子空间 $W$ 的一个**无限**线性无关子集。这是否证明 $W$ 是无限维的？说明理由。

**解答**：

取
$$\{\,S^j(\delta):\ j=0,1,2,\dots\,\}\subseteq W$$

任取其中有限个 $S^{j_1}(\delta),\dots,S^{j_p}(\delta)$（$j_1<\cdots<j_p$）。若
$$c_1S^{j_1}(\delta)+\cdots+c_pS^{j_p}(\delta)=\{0\},$$
则$c_i=0$，所有系数为零，故这有限向量组线性无关

由“无限集线性无关 $\iff$ 每个有限子集无关”，该无限集线性无关。

这证明了 $W$ 无限维。因为： 

若 $W$ 有限维、$\dim W=d$，则$W$ 中任何多于 $d$ 个向量的组都线性相关

但上面的集合含任意大的有限无关子组，矛盾。
故$$\dim W=\infty$$

**题目**：斐波那契数列（4.7 节表 1）满足齐次差分方程
$$y_{k+2}-y_{k+1}-y_k=0,\qquad y_0=0,\ y_1=1.$$
求该数列的通解。

**解答：**

辅助方程 $r^2-r-1=0$ 的根为
$$\varphi=\frac{1+\sqrt5}{2},\qquad \psi=\frac{1-\sqrt5}{2}$$
通解
$$y_k=c_1\varphi^k+c_2\psi^k$$
代入初值：
$$\begin{cases}c_1+c_2=0\\ c_1\varphi+c_2\psi=1\end{cases}
\;\Longrightarrow\;
c_1=-c_2=\frac{1}{\varphi-\psi}=\frac{1}{\sqrt5}$$
故
$$y_k=\frac{1}{\sqrt5}\left[\left(\frac{1+\sqrt5}{2}\right)^k-\left(\frac{1-\sqrt5}{2}\right)^k\right]$$
通解为 $\{y_k\}$

### 习题 4.8 · 第 35 题

**题目**：设 $y_k=k^2$，$z_k=2k|k|$。信号 $\{y_k\}$ 与 $\{z_k\}$ 线性无关吗？对 $k=0,-1,-2$ 计算相应的 Casorati 矩阵 $C(k)$，并讨论你的结果。

**解答：**

```python
for k in [[-2,-1],[-1,0],[0,1]]:
    k_arr = np.array(k)          
    y = k_arr**2
    z = 2 * k_arr * np.abs(k_arr)
    Ck = np.column_stack([y, z])
    print(f"k = {k}")
    print(f"Ck = \n{Ck}")
    print(f"det_Ck = {np.linalg.det(Ck)}\n")
```

【验证：$\{y_k\}$ 与 $\{z_k\}$ 线性无关】

存在 $c_1,c_2$ 对一切 $k$，下列式子成立：

$$c_1k^2 + c_22k|k|=0$$

取 $k=1,-1$

$$c_1 + 2c_2=0$$
$$c_1-2c_2 = 0$$

得 $$c_1 = c_2 = 0$$

所以 $\{y_k\}$ 与 $\{z_k\}$ 线性无关

Casorati 判别法：“某个 $k$ 处 $C(k)$ 可逆 $\Rightarrow$ 线性无关”

本题：信号线性无关，但三个 Casorati 矩阵都不可逆。

所以“ $C(k)$ 奇异”不能推出线性相关

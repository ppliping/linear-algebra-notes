# 第一部分　知识点总结


## 3.1　行列式的引入

- **余子式与代数余子式**：删去 $A$ 的第 $i$ 行第 $j$ 列得到的 $(n-1)\times(n-1)$ 矩阵记为 $A_{ij}$，则
  - **余子式（minor）**：$M_{ij}=\det A_{ij}$；
  - **代数余子式（cofactor）**：$C_{ij}=(-1)^{i+j}\det A_{ij}$（带符号的余子式）。
- **$n\times n$ 行列式（determinant）的定义**（沿第一行的余因子展开，cofactor expansion across the first row of A）：
  $$\det A=a_{11}C_{11}+a_{12}C_{12}+\cdots+a_{1n}C_{1n}.$$
  $1\times1$ 时 $\det[a]=a$。行列式只对**方阵**定义。
- **定理 1**：行列式可沿**任一行或任一列**展开：
  $$\det A=\sum_{j=1}^{n}a_{ij}C_{ij}\ \text{（沿第 } i \text{ 行）},\qquad
    \det A=\sum_{i=1}^{n}a_{ij}C_{ij}\ \text{（沿第 } j \text{ 列）}.$$
  计算策略：挑 **0 最多的行或列** 展开。
- **定理 2**：若 $A$ 是**三角形矩阵**，则 $\det A$ 等于主对角线元素之积。
  - 对角矩阵、单位阵：$\det I=1$。
- **检验算出的行列式是否"合理"**：$-np \le \det A \le np$ ，其中 $A$ 是 $n×n$ 的矩阵， $p$ 是绝对值排名前 $n$ 的元素的乘积（如果某个数值作为矩阵元素出现多次，则该数值重复计入）
- 用代数余子式的方法计算行列式一般要 $n!$ 的计算量，是非常巨大的，实际应用中有更快的算法

## 3.2　行列式的性质

- **定理 3（初等行变换对行列式的作用）**：设 $A$ 为方阵，
  - a. **倍加**：把某行的倍数加到另一行，行列式**不变**
  - b. **行交换**：两行互换，行列式**变号**
  - c. **倍乘**：某行乘 $k$，行列式**乘 $k$**
- **按行化简计算行列式的策略（计算机通常用此法）**：用倍加（不改变）与行交换（变号）把 $A$ 化为阶梯形 $U$：
  $$\det A=(-1)^{r}\cdot(U\text{ 的主元之积})\qquad(r=\text{行交换次数}).$$
- **定理 4（可逆性判据）**：方阵 $A$ 可逆 $\Leftrightarrow$ $\det A\neq0$。
- **定理 5**：$\det A^{T}=\det A$。于是**定理 3 对列同样成立**（列交换变号、列倍乘、列倍加）。
- **定理 6（乘法性质）**：对同阶方阵 $A,B$：
  $$\det AB=(\det A)(\det B).$$
  - 重要推论：
    - $\det A^{-1}=\dfrac{1}{\det A}$（$A$ 可逆时）；
    - $\det A^{k}=(\det A)^{k}$；
    - $\det(rA)=r^{n}\det A$（$A$ 为 $n\times n$，每行都乘 $r$）；$\det(-A)=(-1)^{n}\det A$；
    - $\det(A+B)\neq\det A+\det B$（一般不成立，无加法性质）。
- **对单个列的线性性质**：固定其余列、把第 $j$ 列换成向量变量 $\mathbf x$，则
  $$T(\mathbf x)=\det[\mathbf a_1\ \cdots\ \mathbf a_{j-1}\ \mathbf x\ \mathbf a_{j+1}\ \cdots\ \mathbf a_n]$$
  定义了 $\mathbb R^n\to\mathbb R$ 的**线性变换**：
  - $T(c\mathbf x)=cT(\mathbf x)$（某列乘 $c$，行列式乘 $c$）
  - $T(\mathbf u+\mathbf v)=T(\mathbf u)+T(\mathbf v)$（沿第 $j$ 列余因子展开可证）。
  
  即行列式**对每个列分别线性**，但线性只对**单个列**（其余列固定）成立，对整个矩阵没有加法——$\det(A+B)\neq\det A+\det B$。
- 计算机程序对一般矩阵用**行化简**（$\det A=(-1)^{r}\times$ 主元之积）计算行列式，约需 $2n^{3}/3$ 次算术运算


## 3.3　克拉默法则、体积与线性变换

- **定理 7（克拉默法则 Cramer's Rule）**：设 $A$ 为可逆 $n\times n$ 矩阵，$\mathbf b\in\mathbb R^n$，则 $A\mathbf x=\mathbf b$ 的唯一解为
  $$x_i=\frac{\det A_i(\mathbf b)}{\det A},\qquad i=1,2,\dots,n.$$
  $A_i(\mathbf b)$ 表示把 $A$ 的第 $i$ 列换成 $\mathbf b$ 所得的矩阵。
- **伴随矩阵与逆矩阵公式**：
  - **伴随矩阵（adjugate）** $\operatorname{adj}A$：代数余子式矩阵的**转置**（$(i,j)$ 元为 $C_{ji}$）。
  - **定理 8**：$A$ 可逆时
    $$A^{-1}=\frac{1}{\det A}\operatorname{adj}A,\qquad\text{即 }(A^{-1})_{ij}=\frac{C_{ji}}{\det A}.$$
  - 理论价值为主；数值求逆用行化简快得多。
- **定理 9（行列式的几何意义）**：
  - $A$ 为 $2\times2$：$|\det A|$ = 以 $A$ 的**列**为边的平行四边形**面积**；
  - $A$ 为 $3\times3$：$|\det A|$ = 以 $A$ 的列为棱的长方体**体积**；
  - 符号记录**定向**；三角形面积 $=\frac12|\det[\mathbf u\ \mathbf v]|$。
- **定理 10（线性变换对面积/体积的缩放）**：设 $T:\mathbb R^2\to\mathbb R^2$ 由 $2\times2$ 矩阵 $A$ 确定，$S$ 为平行四边形（可推广到**任一有界区域**），则
  $$\{T(S)\text{ 的面积}\}=|\det A|\cdot\{S\text{ 的面积}\};$$
  $3\times3$ 情形对体积同样成立。

  例子：椭圆 $\dfrac{x_1^2}{a^2}+\dfrac{x_2^2}{b^2}=1$ 是单位圆盘在对角阵 $\operatorname{diag}(a,b)$ 下的像，围成的面积 $=ab\cdot\pi$。

### 习题 3.1 · 第 43 题

**题目**：设 $\mathbf u=\begin{bmatrix}3\\0\end{bmatrix}$，$\mathbf v=\begin{bmatrix}1\\2\end{bmatrix}$。计算由 $\mathbf u,\mathbf v,\mathbf u+\mathbf v,\mathbf 0$ 确定的平行四边形的面积，并计算 $\det[\mathbf u\ \mathbf v]$。两者如何比较？把 $\mathbf v$ 的第一个分量换成任意数 $x$，重复上述计算。画图并解释你的发现。


**解答：**

(a)

平行四边形面积 $$S = 2×3 = 6$$

$$\det[u\, v] =\det \begin{bmatrix}3&1\\0&2\end{bmatrix} = 6-0 = 6$$

$$S=det[u\, v]$$

(b)

$$v = \begin{bmatrix}x\\0\end{bmatrix}$$

平行四边形面积 $$S = 2×3 = 6$$

$$\det[u\, v] =\det \begin{bmatrix}3&x\\0&2\end{bmatrix} = 6-0 = 6$$

$$S=det[u\, v]$$

发现：无论 $v$ 的第一个分量取什么值，都不影响平行四边形面积以及 $det[u\,v]$ 的结果，并且二者总是相等。

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6.5, 4))
# plt.subplots() 创建一个 Figure（画布）和一个 Axes（坐标轴区域）
# figsize=(6.5, 4) 设置画布大小为 6.5×4 英寸；
# 后面所有绘图都通过 ax.xxx() 操作，这是 matplotlib 面向对象接口的写法

for x, c in zip([-2, 0, 1, 4], ['#d62728', '#1f77b4', '#2ca02c', '#9467bd']): # x:v的第一个分量；c:颜色；zip把两组值一一配对
    P = np.array([[0,0], [3,0], [3+x,2], [x,2]]) # 四个顶点
    ax.add_patch(plt.Polygon(P, closed=True, alpha=0.25, color=c, label=f'v = ({x}, 2)'))  #plt.Polygon创建多边形，ax.add_patch 都叠在ax坐标轴区域上
    ax.annotate('u', (1.5, -0.28), fontsize=12, ha='center')  # 在底边中点下方写文字 "u"，水平居中
    ax.annotate('u+v', (3+x+0.15, 2), fontsize=10, color=c) 
    
ax.axhline(2, color='gray', lw=0.5, ls='--') # 在 y=2 处画一条灰色、线宽 0.5 的虚线
ax.plot([0,3,3+x,x], [0,0,2,2], 'k.', ms=4) # 标记所有顶点（x、y坐标配对），'k.'黑色小点，ms=4 点的大小
ax.set_xlim(-3.8, 8.4); ax.set_ylim(-0.8, 2.7) # 设置坐标范围
ax.set_aspect('equal'); # 强制 x 轴和 y 轴单位长度在屏幕上一样长
ax.grid(alpha=0.3); # 设置网格及透明度
ax.legend(loc='upper left', fontsize=7) # 设置图例
plt.show()  # 显示图形
```

### 习题 3.1 · 第 44 题

**题目**：设 $\mathbf u=\begin{bmatrix}a\\b\end{bmatrix}$，$\mathbf v=\begin{bmatrix}c\\0\end{bmatrix}$，其中 $a,b,c$ 为正数（简单起见）。计算由 $\mathbf u,\mathbf v,\mathbf u+\mathbf v,\mathbf 0$ 确定的平行四边形的面积，并计算 $\det[\mathbf u\ \mathbf v]$ 与 $\det[\mathbf v\ \mathbf u]$。画图并解释你的发现。


**解答：**

平行四边形面积 $$S = bc$$

$$\det[u\, v] =\det \begin{bmatrix}a&c\\b&0\end{bmatrix} = -bc$$

$$\det[v\, u] =\det \begin{bmatrix}c&a\\0&b\end{bmatrix} = bc$$

$$S = \det[v\,u] = -\det[u\,v]$$

发现：平行四边形的面积等于由两个向量构成的矩阵的行列式的绝对值，列向量的顺序会影响符号的正负。

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6.5, 4))

a,b,c = 2,3,4
u = np.array([a,b])
v = np.array([c,0])
sum_uv = u+v

# 定义矩阵
uv = np.column_stack([u, v])
vu = np.column_stack([v, u])

# 求行列式
det_uv = np.linalg.det(uv)
det_vu = np.linalg.det(vu)

# 画图
P = np.array([[0,0], v, sum_uv, u]) # 四个顶点
ax.add_patch(plt.Polygon(P, closed=True, alpha=0.25, color='#d62728') ) #plt.Polygon创建多边形，ax.add_patch 都叠在ax坐标轴区域上
ax.annotate('v', (c, -0.28), fontsize=12) 
ax.annotate('u', (a, b+0.28), fontsize=12) 
ax.annotate('u+v', (a+c+0.1, b+0.1), fontsize=10) 

# 设置规格
ax.plot([0,a,a+c,c], [0,b,b,0], 'k.', ms=4) # 标记所有顶点（x、y坐标配对），'k.'黑色小点，ms=4 点的大小
ax.set_xlim(-1, 8.4); ax.set_ylim(-0.8,4) # 设置坐标范围
ax.set_aspect('equal'); # 强制 x 轴和 y 轴单位长度在屏幕上一样长
ax.grid(alpha=0.3); # 设置网格及透明度
plt.show()  # 显示图形

# 输出结果
print(f'平行四边形面积S={b*c}')
print(f'det[u v] = {det_uv}')
print(f'det[v u] = {det_vu}')
```

### 习题 3.1 · 第 47 题 [T]

**题目**：构造一个元素为 $-9$ 到 $9$ 之间整数的随机 $4\times4$ 矩阵 $A$。$\det A^{-1}$ 与 $\det A$ 有何关系？对 $n=4,5,6$ 的随机 $n\times n$ 整数矩阵进行实验，并提出猜想。


**解答：**

```python
rng = np.random.default_rng(47) #随机种子

for n in [4,5,6]:
    A = sp.Matrix(rng.integers(-9, 10, size=(n, n)).tolist()) # 上界不包含，所以范围是 [-9,9]；
    #.tolist() 整数NumPy 数组里的 np.int64 被转成 Python 原生 int；结构变成“每行一个列表”的嵌套列表；这是 SymPy Matrix 最标准、最安全的输入格式
    A_inv = A.inv()
    det_A = A.det()
    det_A_inv = A_inv.det()
    print(f'A = {A}')
    print(f'detA ={det_A}')
    print(f'detA^(-1) = {det_A_inv}')
    print(f'1/detA^(-1) = {1/det_A_inv}\n')
```

猜想：$\det A$ 和 $\det A^{-1}$ 是互为倒数的关系

**解答：**

```python
rng = np.random.default_rng(48)

# 本证明题要用sp
for n in [5,3,7]:
    A = sp.Matrix(rng.integers(-9, 10, size=(n, n)).tolist()) 
    B = sp.Matrix(rng.integers(-9, 10, size=(n, n)).tolist()) 
    AB = A@B
    print(f'n = {n}')
    det_A = A.det();print(f'detA = {det_A}')
    det_B = B.det();print(f'detB = {det_B}')
    det_AB = AB.det();print(f'detAB = {det_AB}')
    print("detAB = (detA)(detB)?",det_AB == det_A*det_B,"\n")
```

### 习题 3.2 · 第 37 题

**题目**：证明若 $A$ 可逆，则 $\det A^{-1}=\dfrac{1}{\det A}$。


**证明：**

因为 $A$ 可逆，所以 $$AA^{-1} = I$$

由定理6，$$\det(AA^{-1}) = (\det A)(\det A^{-1}) = \det I = 1$$

由定理4，$A$ 可逆 $\Leftrightarrow$ $\det A \neq 0$

所以 $$\det A^{-1}=\dfrac{1}{\det A}$$

由定理6，$$\det A^3 = (\det A)(\det A)(\det A) = 0$$ $$\det A = 0$$

由定理4，$\det A =0 \Rightarrow A$ 不可逆

### 习题 3.2 · 第 39 题

**证明：**

这和 $AB\,,BA$ 是否相等无关

### 习题 3.2 · 第 40 题

**题目**：设 $A$ 与 $P$ 为方阵且 $P$ 可逆。证明 $\det(PAP^{-1})=\det A$。


**证明：**

由定理6，$$\det( PAP^{-1} )= (\det P)(\det A)(\det P^{-1}) = (\det P)(\det P^{-1})(\det A) = [\det(PP^{-1})] (\det A)$$

因为 $P$ 可逆，所以 $$PP^{-1} = I$$

所以 $$\det( PAP^{-1}) = (\det I)(\det A) = \det A$$

### 习题 3.2 · 第 41 题

**题目**：设 $U$ 是满足 $U^{T}U=I$ 的方阵。证明 $\det U=\pm1$。


**证明：**

由定理6，$$\det (U^{T}U) = (\det U^T)(\det U) = \det I = 1$$
由定理5，$$\det U = \det U^{T}$$
所以 $$\det (U^{T}U) = (\det U)^2 = 1$$ 
$$\det U = \pm 1$$

### 习题 3.2 · 第 42 题

**题目**：当 $A$ 是 $n\times n$ 矩阵时，求 $\det(rA)$ 的公式。


**解答：**

$rA$ 表示对 $A$ 的每一行都乘 $r$

### 习题 3.2 · 第 52 题

**题目**：设 $A$、$B$ 为 $n\times n$ 矩阵，计算机算得 $\det A=5$、$\det B=2$、$\det AB=7$。应该相信这些结果吗？为什么？

$$A=\begin{bmatrix}4&0&-7&-7\\-6&1&11&9\\7&-5&10&19\\-1&2&3&-1\end{bmatrix}$$

做实验：计算 $A$、$10A$、$0.1A$ 的行列式；对比计算这些矩阵的**条件数**。再用 $4\times4$ 单位矩阵重复上述计算，并讨论你的结果。


**解答：**

```python
print('%-8s %14s %14s' % ('矩阵', 'det', 'cond'))

A = np.array([
    [4,0,-7,-7],
    [-6,1,11,9],
    [7,-5,10,19],
    [-1,2,3,-1]
])

for i in [1,10,0.1]:
    print('%-8s %14.4g %14.4g' % ('%.1fA' %i, np.linalg.det(i*A), np.linalg.cond(i*A)))

I = np.eye(4)

for i in [1,10,0.1]:
    print('%-8s %14.4g %14.4g' % ('%.1fI' %i, np.linalg.det(i*I), np.linalg.cond(i*I)))
```

$A\,,10A\,,0.1A$ 的行列式的大小不相同，但条件数都相同，可见行列式的大小不决定条件数。

$rA\,,rI(r = 1,10,0.1)$ 的条件数才是反映奇异/非奇异的量。条件数大，矩阵接近奇异。

### 习题 3.3 · 第 31 题

**题目**：设 $T:\mathbb R^{3}\to\mathbb R^{3}$ 是由矩阵 $A=\begin{bmatrix}a&0&0\\0&b&0\\0&0&c\end{bmatrix}$（$a,b,c$ 为正数）确定的线性变换，$S$ 是单位球（边界曲面为 $x_1^{2}+x_2^{2}+x_3^{2}=1$）。
a. 证明 $T(S)$ 由椭球面 $x_1^{2}/a^{2}+x_2^{2}/b^{2}+x_3^{2}/c^{2}=1$ 围成；
b. 利用单位球的体积为 $4\pi/3$，求椭球所围区域的体积。


**证明：**

(a)

设 $\mathbf u = [u_1\,u_2\,u_3]^T \, \mathbf x = [x_1\,x_2\,x_3]^T$
$$\mathbf x= T(\mathbf u) = A\mathbf u$$
$$\begin{bmatrix}x1\\x2\\x3\end{bmatrix} = \begin{bmatrix}a&0&0\\0&b&0\\0&0&c\end{bmatrix} \begin{bmatrix}u1\\u2\\u3\end{bmatrix}$$
$$u1=x_1/a \quad u2=x_2/b \quad u3=x_3/c$$
对于每一个在 $u_1^2+u_2^2+u_3^2=1$ 的点，代入上式得 $$x_1^{2}/a^{2}+x_2^{2}/b^{2}+x_3^{2}/c^{2}=1$$

椭球围成的体积为 $abc \tfrac{4\pi}3$

### 习题 3.3 · 第 39 题 [T]

**题目**：用随机 $4\times4$ 矩阵 $A$ 检验定理 8 的逆矩阵公式。用矩阵程序计算各 $3\times3$ 子矩阵的余因子，构造伴随矩阵，并令 $B=(\operatorname{adj}A)/\det A$。再计算 $B-\operatorname{inv}(A)$，其中 $inv(A)$ 是矩阵程序算出的 $A^{-1}$。用可显示最多位小数的浮点算术计算，并报告你的结果。


**解答：**

```python
rng = np.random.default_rng(39)

A = rng.integers(-9,10,size=(4,4)).astype(float)
cof = np.zeros((4,4))  # 余子式矩阵
B = np.zeros((4,4)) 

for i in range(4):
    for j in range(4):
        A_ij = np.delete(np.delete(A,i,axis=0),j,axis=1)
        M_ij = np.linalg.det(A_ij)
        C_ij = (-1) ** ( (i+1)+(j+1) ) * M_ij 
        cof[i,j] = C_ij

adj_A = np.transpose(cof) # 伴随矩阵
B = adj_A / np.linalg.det(A)
inv_A = np.linalg.inv(A)

print(f'A = {A}\n')
print(f'B = {B}\n')
print(f'B - inv(A) = {B - inv_A}\n')
print( 'max |B − inv(A)| =', np.max(np.abs(B - inv_A)) )
```

$B - inv(A)$ 的最大误差在 $10^{-16}$ 量级，这是计算机 float 64 的精度，定理 8 的公式与数值求逆在机器精度内一致

### 习题 3.3 · 第 40 题 [T]

**题目**：用随机 $4\times4$ 矩阵 $A$ 和随机 $4\times1$ 向量 $\mathbf b$ 检验克拉默法则。计算 $A\mathbf x=\mathbf b$ 的解的每个分量，并与 $A^{-1}\mathbf b$ 的对应分量比较。写出用克拉默法则产生解的第二个分量 $x_2$ 的矩阵程序命令。


**解答：**

```python
rng = np.random.default_rng(40)

A = rng.integers(-9,10,size=(4,4)).astype(float)
b = rng.integers(-9,10,size=4).astype(float)

x_np = np.linalg.solve(A,b) 
x_invAb = np.linalg.inv(A) @ b

A2 = A.copy()  # 如果写 A2=A，是浅拷贝，改变 A2 的时候也会改变 A
A2[:,1] = b
x2_cramer = np.linalg.det(A2) / np.linalg.det(A)

print("A = ");print(A)
print(f'\nb = {b}\n')
print("A2 = ");print(A2)
print(f'\nx_np = {x_np}\n')
print(f'x_A^(-1)b = {x_invAb}\n')
print(f'x2_cramer = {x2_cramer}\n')
```

三种方法求出来的结果在浮点数精度下是一致的
